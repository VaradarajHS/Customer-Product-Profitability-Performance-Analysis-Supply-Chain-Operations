```python
required_columns = [
    'Sales',
    'Profit',
    'Discount']
for col in required_columns:
    if col in df.columns:
        print(f'{col}: Available')
    else:
        print(f'{col}: NOT FOUND')
```

```python
if 'Order Date' in df.columns:
    df['Order Date'] = pd.tp_datetime(
        df['Order Date'],
        errors='coerce')
```

```python
df['Profit'] = ( df['Sales'] * df['Order Item Profit Ratio'] )
```

```python
df['Profit Margin (%)'] = (df['Profit'] / df['Sales'] ) * 100
```

```python
total_revenue = df['Sales'].sum()
total_profit = df['Profit'].sum()
profit_margin = ( total_profit / total_revenue) * 100
print(f'Total Revenue: {total_revenue:,.2f}')
print(f'Total Profit: { total_profit:,.2f}')
print(f'Overall Profit Margin: {profit_margin:.2f}%')
```

```python
kpi_summary = pd.DataFrame({
    'KPI':[
    'Total Revenue',
    'Total Profit',
    'Profit Margin(%)',
    'Average Order Value',
    'Total Orders'],
    'Value': [
        df['Sales'].sum(),
        df['Profit'].sum(),
        df['Profit Margin (%)'].sum(),
        df['Sales'].sum() / df['Order Customer Id'].nunique(),
        df['Order Customer Id'].nunique()]
})
display(kpi_summary)
        
```

```python
metrics = [
    'Revenue',
    'Profit']
values = [ df['Sales'].sum(),
           df['Profit'].sum()]
plt.figure(figsize=(9,6))
plt.bar(metrics, values)
plt.title('Revenue vs Profit',
          fontsize=18,
          fontweight='bold')
plt.ylabel('Amount')
plt.tight_layout()
plt.show()
```

```python
category_analysis = (
    df.groupby('Category Name')
    .agg(
        Revenue=('Sales', 'sum'),
        Profit=('Profit', 'sum')
    )
    .reset_index()
)
category_analysis['Profit Margin (%)'] = (
    category_analysis['Profit'] /
    category_analysis['Revenue']
) * 100
category_analysis = category_analysis.sort_values(
    'Profit',
    ascending=False
)
display(category_analysis)
```

```python
plt.figure(figsize=(10, 6))
plt.bar(
    category_analysis['Category Name'],
    category_analysis['Profit'])
plt.title(
    'Profit by Product Category',
    fontsize=18,
    fontweight='bold')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
product_analysis = (
    df.groupby('Product Name')
    .agg(
        Revenue=('Sales', 'sum'),
        Profit=('Profit', 'sum'))
    .reset_index()
)
product_analysis['Profit Margin (%)']=(
    product_analysis['Profit']/ 
    product_analysis['Revenue'])* 100
```

```python
revenue_threshold = product_analysis['Revenue'].quantile(0.75)
low_margin_threshold = 10
high_revenue_low_margin = product_analysis[
    (product_analysis['Revenue']>= revenue_threshold) &
    (product_analysis['Profit Margin (%)'] < low_margin_threshold)]
display(
    high_revenue_low_margin.sort_values(
        'Revenue',
        ascending=False
    ).head(20)
)
```

```python
customer_analysis = (
    df.groupby('Customer Id')
    .agg(
        Revenue=('Sales', 'sum'),
        Profit=('Profit', 'sum'),
              Orders=('Customer Id', 'count')
    )
    .reset_index()
)
customer_analysis['Profit Margin (%)'] = (
    customer_analysis['Profit']/
    customer_analysis['Revenue'])*100
display(customer_analysis.head())
```

```python
high_value_threshold = customer_analysis["Profit"].quantile(0.80)

high_value_customers = customer_analysis[
    customer_analysis["Profit"] >= high_value_threshold
]

display(
    high_value_customers.sort_values(
        "Profit",
        ascending=False
    ).head(20)
)
```

```python
low_margin_customers = customer_analysis[
    customer_analysis['Profit Margin (%)'] < 5]
display(
    low_margin_customers.sort_values(
        'Profit Margin (%)'
    ).head(20))
```

```python
loss_making_customers= customer_analysis[
    customer_analysis['Profit']<0
    ]
display(
    loss_making_customers.sort_values(
        'Profit').head())
```

```python
customer_analysis['Value Tier'] = pd.qcut(
    customer_analysis['Profit'],
    q=4,
    labels=[
        'Low Value',
        'Medium Value',
        'High Value',
        'Very High Value'])
```

```python
customer_analysis['Value Tier'].value_counts()
```

```python
tier_counts = (
    customer_analysis['Value Tier']
    .value_counts()
    .sort_index())
plt.figure(figsize=(10,6))
plt.bar(
    tier_counts.index.astype(str),
    tier_counts.values)
plt.title(
    'Customer Distribution by Value Tier',
    fontsize=18,
    fontweight='bold')
plt.xlabel('Customer Value Tier')
plt.ylabel('Number of Customers')
```

```python
df['Original Value'] = (df['Order Item Product Price'] * df['Order Item Quantity'])
```

```python
df['Discount Amount']=(df['Original Value'] - df['Order Item Total'])
```

```python
df['Discount'] = (df['Discount Amount'] / df['Original Value'] ) * 100
```

```python
df['Discount Group'] = pd.cut(
    df['Discount'],
    bins=[-0.01,0,0.10,0.20,0.30,0.40,1.00],
    labels=[
        'No Discount','1-10%','11-20%','21-30','31-40%','40%+']
)
```

```python
discount_analysis = (
    df.groupby('Discount Group', observed=False)
    .agg(
        Revenue=('Sales', 'sum'),
        Profit=('Profit', 'sum'),
        Average_Discount=('Discount', 'mean')
    )
    .reset_index()
)
discount_analysis['Profit Margin (%)'] = (
    discount_analysis['Profit'] /
    discount_analysis['Revenue']) * 100
display(discount_analysis)
```

```python
plt.figure(figsize=(11, 6))
plt.plot(
    discount_analysis['Discount Group'].astype(str),
    discount_analysis['Profit Margin (%)'],
    marker='o'
)
plt.title(
    'Discount Rate vs Profit Margin',
    fontsize=18,
    fontweight="bold"
)
plt.xlabel('Discount Group')
plt.ylabel('Profit Margin (%)')
plt.xticks(rotation=30)
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
df['Discount Status'] = np.where(
    df['Discount'] > 0,
    'Discounted',
    'No Discount')
```

```python
discount_comparison = (
    df.groupby('Discount Status')
    .agg(
        Revenue=('Sales', 'sum'),
        Profit=('Profit', 'sum'),
        Average_Margin=('Profit Margin (%)', 'mean')
    )
    .reset_index()
)
display(discount_comparison)
```
